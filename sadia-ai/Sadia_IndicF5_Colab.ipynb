```
from google.colab import files
uploaded = files.upload()
reference_path = next(iter(uploaded))
assert reference_path.lower().endswith('.wav'), 'Upload a WAV reference'
ref_text = input('Exact transcript of the WAV recording: ').strip()
assert ref_text, 'Transcript is required'
text = 'হ্যালো, আমি সাদিয়া। তোমার সঙ্গে কথা বলতে ভালো লাগছে।'
```

```
from transformers import AutoModel
import numpy as np, soundfile as sf
model = AutoModel.from_pretrained('ai4bharat/IndicF5', trust_remote_code=True)
audio = np.asarray(model(text, ref_audio_path=reference_path, ref_text=ref_text))
if audio.dtype == np.int16: audio = audio.astype(np.float32) / 32768.0
sf.write('sadia-test.wav', audio.astype(np.float32), 24000)
from IPython.display import Audio, display
display(Audio('sadia-test.wav'))
```
